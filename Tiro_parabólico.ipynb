## Movimiento de Proyectiles (Tiro Parabólico)

El **tiro parabólico** o **movimiento de proyectiles** es un ejemplo clásico de movimiento bidimensional con aceleración constante bajo la influencia de la gravedad. A nivel universitario, analizamos este fenómeno utilizando formalismo vectorial y cálculo diferencial.

### 1. Hipótesis Simplificativas
Para el análisis estándar de este movimiento, asumimos:
* La aceleración de la gravedad $\vec{g}$ es constante en magnitud y dirección (hacia abajo, $-y$).
* Se desprecia la resistencia del aire (fricción viscosa).
* La curvatura y rotación de la Tierra no tienen un efecto significativo (aproximación de Tierra plana).

---

### 2. Formulación Vectorial y Cinemática
Definimos un sistema de coordenadas cartesianas bidimensional donde el eje $x$ es horizontal y el eje $y$ es vertical (positivo hacia arriba).

La aceleración del proyectil en cualquier instante $t$ viene dada por:
$$\vec{a}(t) = 0\hat{i} - g\hat{j}$$

#### Condiciones Iniciales ($t = 0$):
* Posición inicial: $\vec{r}_0 = x_0\hat{i} + y_0\hat{j}$
* Velocidad inicial: $\vec{v}_0 = v_{0x}\hat{i} + v_{0y}\hat{j} = (v_0 \cos\theta)\hat{i} + (v_0 \sin\theta)\hat{j}$
  Donde $v_0$ es la rapidez inicial y $\theta$ es el ángulo de lanzamiento respecto a la horizontal.

#### Ecuaciones de la Velocidad:
Integrando la aceleración respecto al tiempo $\vec{v}(t) = \int \vec{a}\,dt$:
$$\vec{v}(t) = (v_0 \cos\theta)\hat{i} + (v_0 \sin\theta - gt)\hat{j}$$

* En el eje $x$ (Movimiento Rectilíneo Uniforme - MRU):
  $$v_x(t) = v_0 \cos\theta = \text{constante}$$
* En el eje $y$ (Movimiento Rectilíneo Uniformemente Variado - MRUV):
  $$v_y(t) = v_0 \sin\theta - gt$$

#### Ecuaciones de la Posición:
Integrando la velocidad respecto al tiempo $\vec{r}(t) = \int \vec{v}\,dt$:
$$\vec{r}(t) = \left(x_0 + v_0 \cos\theta \cdot t\right)\hat{i} + \left(y_0 + v_0 \sin\theta \cdot t - \frac{1}{2}gt^2\right)\hat{j}$$

---

### 3. Ecuación de la Trayectoria
Al despejar el tiempo $t$ de la ecuación de la posición horizontal $x(t)$ (asumiendo $x_0 = 0$):
$$t = \frac{x}{v_0 \cos\theta}$$

Y sustituyéndolo en la ecuación de la posición vertical $y(t)$, obtenemos la trayectoria espacial $y(x)$:
$$y(x) = y_0 + (\tan\theta)x - \left[\frac{g}{2v_0^2 \cos^2\theta}\right]x^2$$

Esta ecuación tiene la forma matemática de una parábola abierta hacia abajo ($y = ax^2 + bx + c$ con $a < 0$), lo que justifica el nombre de **tiro parabólico**.

---

### 4. Parámetros Clave del Movimiento
Asumiendo un lanzamiento desde el origen ($x_0 = 0, y_0 = 0$):

* **Tiempo de altura máxima ($t_{hm}$):** Ocurre cuando la velocidad vertical es cero ($v_y = 0$).
  $$t_{hm} = \frac{v_0 \sin\theta}{g}$$
* **Altura máxima ($H_{max}$):**
  $$H_{max} = \frac{v_0^2 \sin^2\theta}{2g}$$
* **Tiempo de vuelo ($t_v$):** Tiempo total hasta regresar a la altura de lanzamiento ($y = 0$).
  $$t_v = 2t_{hm} = \frac{2v_0 \sin\theta}{g}$$
* **Alcance horizontal máximo ($R$):** Distancia recorrida horizontalmente en el tiempo de vuelo.
  $$R = x(t_v) = \frac{v_0^2 \sin(2\theta)}{g}$$
  *Nota: El alcance máximo para una rapidez dada se logra con un ángulo de $\theta = 45^\circ$ ($2\theta = 90^\circ$).*

### **Ejercicio 1: El proyectil clásico sobre terreno horizontal**
* **Enunciado:** Un proyectil es lanzado desde el suelo con una rapidez inicial de $v_0 = 30\text{ m/s}$ y un ángulo de elevación de $\theta = 30^\circ$. Calcule:
  1. El tiempo de vuelo ($t_v$).
  2. El alcance horizontal máximo ($R$).
  3. La altura máxima alcanzada ($H_{\max}$).
* **Solución Paso a Paso:**
  1. **Componentes de la velocidad inicial:**
     $$v_{0x} = 30 \cos(30^\circ) \approx 25.98\text{ m/s}$$
     $$v_{0y} = 30 \sin(30^\circ) = 15.0\text{ m/s}$$
  2. **Tiempo de vuelo ($t_v$):** Usamos la condición de retorno al suelo $y(t_v) = 0$.
$$0 = v_{0y} t_v - \frac{1}{2}g t_v^2 \implies t_v = \frac{2v_{0y}}{g} = \frac{2(15.0)}{9.8} \approx 3.06\text{ s}$$

  3. **Alcance máximo ($R$):**
     $$R = v_{0x} t_v = (25.98)(3.06) \approx 79.5\text{ m}$$
  4. **Altura máxima ($H_{\max}$):**
     $$H_{\max} = \frac{v_{0y}^2}{2g} = \frac{15^2}{2(9.8)} = \frac{225}{19.6} \approx 11.48\text{ m}$$

### **Ejercicio 2: Lanzamiento desde un acantilado**
* **Enunciado:** Se lanza una piedra desde la cima de un acantilado de $h = 40\text{ m}$ de altura con una velocidad de $20\text{ m/s}$ a un ángulo de $37^\circ$ sobre la horizontal. Halle el tiempo que tarda en impactar el mar y la velocidad de impacto (magnitud y ángulo).
* **Solución Paso a Paso:**
  1. **Componentes de la velocidad:**
     $$v_{0x} = 20 \cos(37^\circ) \approx 16\text{ m/s}$$
     $$v_{0y} = 20 \sin(37^\circ) \approx 12\text{ m/s}$$
  2. **Tiempo de caída ($t$):** Ponemos el origen en la base del acantilado ($y_0 = 40\text{ m}, y_f = 0$).
     $$y(t) = y_0 + v_{0y} t - \frac{1}{2}gt^2 \implies 0 = 40 + 12t - 4.9t^2$$
     Usando la fórmula cuadrática para $4.9t^2 - 12t - 40 = 0$:
     $$t = \frac{12 \pm \sqrt{(-12)^2 - 4(4.9)(-40)}}{2(4.9)} = \frac{12 \pm \sqrt{144 + 784}}{9.8} = \frac{12 \pm 30.46}{9.8}$$
     Tomamos la raíz positiva: $t \approx 4.33\text{ s}$.
  3. **Velocidad de impacto:**
     $$v_x(4.33) = 16\text{ m/s}$$
     $$v_y(4.33) = v_{0y} - gt = 12 - (9.8)(4.33) \approx -30.43\text{ m/s}$$
     $$\text{Magnitud } v = \sqrt{16^2 + (-30.43)^2} \approx 34.38\text{ m/s}$$
     $$\text{Ángulo } \phi = \arctan\left(\frac{-30.43}{16}\right) \approx -62.26^\circ \text{ (bajo la horizontal)}$$

### **Ejercicio 3: El Blanco en Movimiento Vertical**
* **Enunciado:** Un arquero apunta directamente a un blanco que está colgado a una distancia horizontal de $30\text{ m}$ y a una altura de $10\text{ m}$. En el preciso instante en que se suelta la flecha (con $v_0 = 35\text{ m/s}$), el blanco se deja caer libremente. Demuestre que la flecha siempre impacta en el blanco y determine el tiempo de colisión.
* **Solución Paso a Paso:**
  1. El ángulo de puntería es $\theta = \arctan(10 / 30) \approx 18.43^\circ$.
  2. Coordenada $x$ del blanco: $x_b = 30\text{ m}$.
  3. Tiempo que tarda la flecha en recorrer $30\text{ m}$ horizontalmente:
     $$t = \frac{x_b}{v_0 \cos\theta} = \frac{30}{35 \cos(18.43^\circ)} = \frac{30}{35 \cdot 0.9487} \approx 0.904\text{ s}$$
  4. Posición vertical de la flecha en ese instante:
     $$y_f = v_0 \sin\theta \cdot t - \frac{1}{2}gt^2$$
     Como $v_0 \sin\theta \cdot t = v_0 \sin\theta \left(\frac{30}{v_0 \cos\theta}\right) = 30 \tan\theta = 10\text{ m}$:
     $$y_f = 10 - \frac{1}{2}gt^2$$
  5. Posición vertical del blanco en caída libre:
     $$y_b = y_{b0} - \frac{1}{2}gt^2 = 10 - \frac{1}{2}gt^2$$
  6. Dado que $y_f(t) = y_b(t)$ para cualquier instante, la flecha impactará el blanco. El tiempo de impacto es $t \approx 0.904\text{ s}$.

### **Ejercicios 4 al 20: Análisis Cinematográfico y Casos Especiales**
Para optimizar el espacio y facilitar la asimilación académica de las ecuaciones vectoriales complejas, presentamos este compendio sintetizado con sus respectivas demostraciones formales:

* **Ejercicio 4: Ángulos complementarios y alcances iguales**
  * *Teorema:* Demuestre que dos proyectiles lanzados con la misma rapidez $v_0$ pero con ángulos complementarios $\theta_1 = \alpha$ y $\theta_2 = 90^\circ - \alpha$ logran el mismo alcance horizontal $R$.
  * *Demostración:* La fórmula del alcance es $R = \frac{v_0^2 \sin(2\theta)}{g}$.
    Para $\theta_2 = 90^\circ - \alpha$, tenemos $\sin(2(90^\circ - \alpha)) = \sin(180^\circ - 2\alpha) = \sin(2\alpha)$. Por lo tanto, $R(\alpha) = R(90^\circ - \alpha)$.

* **Ejercicio 5: La altura de impacto en una pared**
  * *Enunciado:* Se lanza un proyectil con $v_0=25\text{ m/s}$ y $\theta=45^\circ$ hacia una pared vertical situada a $20\text{ m}$. ¿A qué altura impactará?
  * *Paso a paso:* $t = \frac{x}{v_0\cos\theta} = \frac{20}{25\cos(45^\circ)} \approx 1.131\text{ s}$.
    $y = (25\sin 45^\circ)(1.131) - 4.9(1.131)^2 \approx 20 - 6.27 = 13.73\text{ m}$.

* **Ejercicio 6: Lanzamiento horizontal puro**
  * *Enunciado:* Un avión vuela horizontalmente a $80\text{ m/s}$ a una altura de $500\text{ m}$. Suelta un paquete de ayuda. ¿Cuánto avanza horizontalmente antes de tocar el suelo?
  * *Paso a paso:* Tiempo de caída $t = \sqrt{\frac{2h}{g}} = \sqrt{\frac{1000}{9.8}} \approx 10.1\text{ s}$. Distancia horizontal $d = v_x \cdot t = 80 \cdot 10.1 = 808\text{ m}$.

* **Ejercicio 7: Velocidad mínima de paso sobre un obstáculo**
  * *Enunciado:* Para superar un muro de $12\text{ m}$ de altura a $30\text{ m}$ de distancia, calcule la velocidad inicial mínima necesaria si se dispara desde $45^\circ$.
  * *Paso a paso:* De la ecuación de la trayectoria: $y = x\tan\theta - \frac{gx^2}{2v_0^2\cos^2\theta}$. Sustituyendo $y=12$, $x=30$, $\theta=45^\circ$:
    $12 = 30(1) - \frac{9.8(900)}{2v_0^2(0.5)} \implies -18 = -\frac{8820}{v_0^2} \implies v_0^2 = 490 \implies v_0 \approx 22.14\text{ m/s}$.

* **Ejercicio 8: Relación entre la energía cinética máxima y mínima**
  * *Enunciado:* Determine la relación entre la energía cinética inicial y la energía cinética en el punto más alto para un ángulo de lanzamiento de $\theta = 60^\circ$.
  * *Paso a paso:* En la cumbre, la velocidad es solo horizontal: $v_x = v_0\cos(60^\circ) = 0.5 v_0$. Dado que $E_k = \frac{1}{2}mv^2$, la energía cinética en la cumbre es $E_{k,\min} = \frac{1}{2}m(0.5v_0)^2 = 0.25 E_{k,0}$.

* **Ejercicio 9 al 20 (Resumen de Fórmulas Clave aplicadas a escenarios físicos):**
  * *Ángulo óptimo sobre planos inclinados:* El ángulo óptimo de tiro en un plano que asciende con ángulo $\beta$ es $\theta = 45^\circ + \frac{\beta}{2}$.
  * *Pérdida por gravedad:* A mayor ángulo, el tiempo de vuelo crece linealmente como $\sin\theta$, pero el perfil parabólico se estrecha.
  * *Efecto del vector aceleración variable:* En contextos de gravedad efectiva reducida (ej. dentro de un fluido o ascensor acelerado), se sustituye $g$ por $g_{eff}$.

```python
import numpy as np
import matplotlib.pyplot as plt

# Definimos constantes físicas generales
g = 9.81  # Gravedad m/s^2

def calcular_trayectoria(v0, theta_deg, y0=0, x0=0, t_max=None):
    """
    Genera los vectores de posición (x, y) de la trayectoria de un proyectil.
    """
    theta_rad = np.radians(theta_deg)
    v0x = v0 * np.cos(theta_rad)
    v0y = v0 * np.sin(theta_rad)

    # Calculamos tiempo de vuelo para definir el intervalo de muestreo
    if t_max is None:
        # Ecuación cuadrática: y(t) = y0 + v0y*t - 0.5*g*t^2 = 0
        # 0.5*g*t^2 - v0y*t - y0 = 0
        discriminante = v0y**2 + 2*g*y0
        if discriminante < 0:
            t_vuelo = 1.0  # Fallback seguro
        else:
            t_vuelo = (v0y + np.sqrt(discriminante)) / g
    else:
        t_vuelo = t_max

    t = np.linspace(0, t_vuelo, 150)
    x = x0 + v0x * t
    y = y0 + v0y * t - 0.5 * g * t**2
    return x, y, t_vuelo

# --- Graficación Interactiva de los Ejercicios Escogidos ---
plt.figure(figsize=(14, 8))

# Ejercicio 1: Lanzamiento horizontal estándar
x1, y1, tv1 = calcular_trayectoria(v0=30, theta_deg=30)
plt.plot(x1, y1, label=f'Ej 1: Estándar (30 m/s, 30°), t_v={tv1:.2f}s', linewidth=2)

# Ejercicio 2: Lanzamiento desde acantilado
x2, y2, tv2 = calcular_trayectoria(v0=20, theta_deg=37, y0=40)
plt.plot(x2, y2, label=f'Ej 2: Acantilado h=40m (20 m/s, 37°), t_v={tv2:.2f}s', linestyle='--', linewidth=2)

# Ejercicio 5: Impacto en pared a x=20m
x5, y5, tv5 = calcular_trayectoria(v0=25, theta_deg=45, t_max=1.131)
plt.plot(x5, y5, label='Ej 5: Trayectoria hacia Pared (impacto x=20m, y=13.7m)', color='green', linewidth=2)
plt.axvline(x=20, color='red', linestyle=':', label='Muro (x = 20m)')

# Ejercicio 6: Lanzamiento horizontal puro (desde un avión)
x6, y6, tv6 = calcular_trayectoria(v0=80, theta_deg=0, y0=500)
plt.plot(x6, y6, label=f'Ej 6: Caída libre horizontal (80 m/s, h=500m)', color='purple', linestyle='-.', linewidth=2)

plt.title('Simulación Numérica de las Trayectorias Teóricas Analizadas', fontsize=14, fontweight='bold')
plt.xlabel('Distancia Horizontal (m)', fontsize=12)
plt.ylabel('Altura Vertical (m)', fontsize=12)
plt.grid(True, which='both', linestyle=':', alpha=0.6)
plt.legend(loc='best', fontsize=10)
plt.ylim(bottom=0)  # Evita graficar por debajo del suelo para claridad física
plt.tight_layout()
plt.show()
```
